```python
import math


def read_instance(path):
    with open(path, encoding="utf-8") as f:
        n, m = map(int, f.readline().split())

        facilities = []
        customers = []

        for _ in range(n):
            cost, capacity, x, y = map(float, f.readline().split())
            facilities.append((capacity, cost, x, y))

        for _ in range(m):
            demand, x, y = map(float, f.readline().split())
            customers.append((demand, x, y))

    return facilities, customers


def greedy_facility(facilities, customers):
    remaining = [f[0] for f in facilities]
    opened = set()
    assignment = [-1] * len(customers)

    avg_demand = sum(c[0] for c in customers) / len(customers)

    for c, (demand, x, y) in enumerate(customers):
        best = None

        for f, (capacity, cost, fx, fy) in enumerate(facilities):
            if demand > remaining[f]:
                continue

            distance = math.dist((x, y), (fx, fy))

            opening = 0 if f in opened else cost * avg_demand / capacity

            score = distance + opening

            if best is None or score < best[0]:
                best = score, f

        _, f = best

        assignment[c] = f
        remaining[f] -= demand
        opened.add(f)

    return opened, assignment


def total_cost(opened, assignment, facilities, customers):
    total = sum(facilities[f][1] for f in opened)

    for c, f in enumerate(assignment):
        total += math.dist(
            facilities[f][2:],
            customers[c][1:]
        )

    return total


tests = [
    "data/fl_25_2",
    "data/fl_100_1",
    "data/fl_200_7",
    "data/fl_500_7",
    "data/fl_1000_2",
    "data/fl_2000_2",
]

for test in tests:
    facilities, customers = read_instance(test)

    opened, assignment = greedy_facility(
        facilities,
        customers
    )

    cost = total_cost(
        opened,
        assignment,
        facilities,
        customers
    )

    print(test, cost)
```
